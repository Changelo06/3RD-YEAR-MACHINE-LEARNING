```python
#1 Discount Checker
num_items = int(input("Enter Number of items"))
price = int(input("Enter Price: "))
total = num_items * price
discount = 0
if num_items*price > 500:
    discount = (num_items * price) * .10
    
print("Total Price:", total) 
print("Discount:", discount)
print("TOTAL:", total-discount)
```

```python
#April-7-2025
quantity = 0
juice = 25
chocolate = 30
cookies = 20
total_cost = 0

while True:
    print("Menu of Items")
    print("1-Juice(" + str(juice) +")")
    print("2-Chocolate(" + str(chocolate) +")")
    print("3-Cookies(" + str(cookies) +")")
    print("4-Exit")

    user_input = int(input("Select an item (1-4): ")

    if user_input == 1:
        print("Juice added (PHP" + str(juice) + ")")
        total_cost += juice
    if user_input == 2:
        print("Chocolate added (PHP" + str(chocolate) + ")")
         total_cost += chocolate
    if user_input == 3:
        print("Cookies added (PHP" + str(cookies) + ")")
         total_cost += cookies
    if user_input < 4 and user_input > 0:
        quantity +=1:
        
    if user_input == 4:
        print("Purchase Summary")
        print("Total items: ", quantity)
        print("Total cost: ", total_cost)
        print("Average cost per item: ", total_cost/quantity)
        break
    else:
        print("Invalid Choice. Try Again!")
    
```
