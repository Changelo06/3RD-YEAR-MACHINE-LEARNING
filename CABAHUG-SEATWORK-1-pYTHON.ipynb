```python
user_data = {}

def create_account():
    global user_data
    username = input("Please input your name: ")

    while True:
        pin = input("Set your 4-digit PIN: ")
        if len(pin) == 4 and pin.isdigit():
            user_data['username'] = username #stores username in user_data
            user_data['pin'] = pin #stores pin in user_data
            print("Account created successfully!")
            break
        else:
            print("Invalid PIN! It must be exactly 4 digits or has no characters!")

def login():
    if not user_data:
        print("No account data found! Please create an account first!")
        return False

    while True:
        username = input("Enter your name: ")
        pin = input("Enter your 4-digit PIN: ")

        if username == user_data.get('username') and pin == user_data.get('pin'): #validates the credentials of the login
            print("Login successful! Welcome, ", username)
            return True
        else:
            print("Invalid credentials! Please try again.")

def transaction_system():
    savings = 0
    
    while True: 
        print("\nTransaction Menu:")
        print("1. Deposit")
        print("2. Withdraw")
        print("3. Exit")

        choice = input("Choose an option (1/2/3): ")

        match choice:
            case "1": #deposits amount into savings
                amount = float(input("Enter amount to deposit: "))
                savings += amount
                print(f"You have deposited: PHP {amount}")
                print(f"Current balance: PHP {savings}")

            case "2":  #withdraw system, subtracts the amount withdrawn to savings. also validates if you have enough amount to withdraw
                if savings <= 0:
                    print("Insufficient balance! You cannot withdraw.")
                    continue
                
                amount = float(input("Enter amount to withdraw: "))
                
                if amount <= savings:
                    savings -= amount
                    print(f"You have withdrawn: PHP {amount}")
                    print(f"Remaining balance: PHP {savings}")
                else:
                    print("Withdrawal amount exceeds your balance!")

            case "3":
                print("Exiting transactions.")
                break
            
            case _:
                print("Invalid option. Please try again.")

    print("\n--- Final Receipt ---")
    print(f"Total savings remaining: PHP {savings}") # prints what is left in the savings after the transactions

def main():
    while True:
        print("\nMenu:\n1. Create Account\n2. Login\n3. Exit")
        choice = input("Choose an option: (1. Create an account | 2. Login | 3. Exit)")

        match choice:
            case "1":
                create_account()
            case "2":
                logged_in = login # stores the condition of the login: true or false
                if login(): 
                    transaction_system()
            case "3":
                print("Exiting System. Goodbye!")
                break
            case _:
                print("Invalid option! Please choose again.")

main()
```

```python
user_data = {}
#Multiple Users
def create_acc():
    global user_data

    while True:
        username = input("Setup Username (Must have exactly 8 characters): ")
        if len(username) != 8:
            print("Username must be exactly 8 characters.")
        elif username in user_data:
            print("Username already exists. Try another one.")
        else:
            break

    while True:
        pin = input("Setup 4-Digit PIN: ")
        if len(pin) == 4 and pin.isdigit():
            break
        else:
            print("PIN must be exactly 4 digits.")

    while True:
        try:
            balance = float(input("Enter Initial Balance (minimum 500): "))
            if balance >= 500:
                break
            else:
                print("Initial balance must be at least 500.")
        except ValueError:
            print("Please enter a valid number for balance.")

    # Store user details in nested dictionary
    user_data[username] = {
        'pin': pin,
        'bal': balance
    }
    print("Account created successfully.")

def login_acc():
    attempts = 3
    if not user_data:
        print("No accounts created yet. Please create an account.")
        return None
    while attempts > 0:
        username = input("Enter Username: ")
        pin = input("Enter 4-Digit PIN: ")
        if username in user_data and user_data[username]['pin'] == pin:
            print("Access granted.")
            return username
        else:
            attempts -= 1
            print("Incorrect username or PIN.")
            print("Attempts left:", attempts)
    print("Too many failed attempts. Returning to main menu.")
    return None

def transaction():
    username = login_acc()
    if username:
        while True:
            print("\nTransaction Menu")
            print("(1) Deposit")
            print("(2) Withdraw")
            print("(3) Back to Main Menu")
            choice = input("Choose an option: ")

            if choice == "1":
                try:
                    amount = float(input("Enter amount to deposit: "))
                    user_data[username]['bal'] += amount
                    print(f"Deposit successful. New balance: {user_data[username]['bal']}")
                except ValueError:
                    print("Invalid amount.")
            elif choice == "2":
                try:
                    amount = float(input("Enter amount to withdraw: "))
                    if amount > user_data[username]['bal']:
                        print("Insufficient funds.")
                    else:
                        user_data[username]['bal'] -= amount
                        print(f"Withdrawal successful. New balance: {user_data[username]['bal']}")
                except ValueError:
                    print("Invalid amount.")
            elif choice == "3":
                print("Returning to main menu.")
                break
            else:
                print("Invalid choice.")

def account_details():
    username = login_acc()
    if username:
        print("\nAccount Details")
        print("Username:", username)
        print("Balance: ", user_data[username]['bal'])

def main():
    while True:
        print("\nWelcome to Cabahug Bank")
        print("(1) Create an Account")
        print("(2) Withdraw/Deposit")
        print("(3) Check Account Details")
        print("(4) Exit")

        try:
            choice = int(input("You: "))
            if choice == 1:
                create_acc()
            elif choice == 2:
                transaction()
            elif choice == 3:
                account_details()
            elif choice == 4:
                print("Exiting system. Goodbye!")
                break
            else:
                print("Invalid input. Please enter 1 to 4.")
        except ValueError:
            print("Please enter a valid number.")

main()
```

```python
def add(x, y):
    return x + y

def subtract(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    if y != 0:
        return x / y
    else:
        return "Cannot divide by zero."

print("Simple Calculator")
print("1. Add\n2. Subtract\n3. Multiply\n4. Divide")

choice = input("Enter choice (1/2/3/4): ")
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

if choice == '1':
    print("Result:", add(num1, num2))
elif choice == '2':
    print("Result:", subtract(num1, num2))
elif choice == '3':
    print("Result:", multiply(num1, num2))
elif choice == '4':
    print("Result:", divide(num1, num2))
else:
    print("Invalid Input")
```

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

number = int(input("Enter a number: "))
if is_prime(number):
    print(number, "is a prime number.")
else:
    print(number, "is not a prime number.")
```

```python
import string

def check_password_strength(password):
    length = len(password) >= 8
    upper = any(c.isupper() for c in password)
    lower = any(c.islower() for c in password)
    digit = any(c.isdigit() for c in password)
    special = any(c in string.punctuation for c in password)

    if all([length, upper, lower, digit, special]):
        return "Strong"
    elif length and (upper or lower) and digit:
        return "Moderate"
    else:
        return "Weak"

pwd = input("Enter your password: ")
print("Password Strength:", check_password_strength(pwd))
```

```python
inventory = {}

def add_item(item, quantity):
    inventory[item] = inventory.get(item, 0) + quantity
    print(f"{quantity} {item}(s) added.")

def remove_item(item, quantity):
    if item in inventory:
        inventory[item] -= quantity
        if inventory[item] <= 0:
            del inventory[item]
            print(f"{item} removed from inventory.")
        else:
            print(f"{quantity} {item}(s) removed.")
    else:
        print("Item not found!")

def show_inventory():
    if inventory:
        print("Inventory:")
        for item, qty in inventory.items():
            print(f"- {item}: {qty}")
    else:
        print("Inventory is empty.")

while True:
    print("\n[1] Add Item\n[2] Remove Item\n[3] Show Inventory\n[4] Exit")
    choice = input("Choose an option: ")
    
    if choice == "1":
        item = input("Item name: ")
        qty = int(input("Quantity: "))
        add_item(item, qty)
    elif choice == "2":
        item = input("Item name: ")
        qty = int(input("Quantity: "))
        remove_item(item, qty)
    elif choice == "3":
        show_inventory()
    elif choice == "4":
        break
    else:
        print("Invalid choice.")
```

```python
def ask_question(question, options, answer):
    print(question)
    for i, option in enumerate(options):
        print(f"{i+1}. {option}")
    choice = int(input("Your answer (1-4): "))
    return options[choice - 1] == answer

score = 0
score += ask_question("What is the capital of France?", ["Berlin", "London", "Paris", "Madrid"], "Paris")
score += ask_question("What language is used for web apps?", ["Python", "HTML", "C++", "Java"], "HTML")

print(f"You got {score}/2 correct.")
```

```python
#TODO LIST WITH APPEND 
todo_list = []

def add_task(task):
    todo_list.append(task)
    print(f"'{task}' added.")

def remove_task(index):
    try:
        removed = todo_list.pop(index)
        print(f"'{removed}' removed.")
    except IndexError:
        print("Invalid task number.")

def show_tasks():
    if not todo_list:
        print("No tasks.")
    else:
        print("To-Do List:")
        for i, task in enumerate(todo_list):
            print(f"{i + 1}. {task}")

while True:
    print("\n[1] Add Task\n[2] Remove Task\n[3] Show Tasks\n[4] Exit")
    choice = input("Choose: ")

    if choice == '1':
        task = input("Enter task: ")
        add_task(task)
    elif choice == '2':
        show_tasks()
        idx = int(input("Enter task number to remove: ")) - 1
        remove_task(idx)
    elif choice == '3':
        show_tasks()
    elif choice == '4':
        break
    else:
        print("Invalid choice.")
```

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
temperatures = []

for day in days:
    while True:
        temp_input = input(f"Enter temperature for {day} (in °C): ")
        try:
            temp = float(temp_input)
            if -50 <= temp <= 60:
                temperatures.append(temp)
                break
            else:
                print("⚠️ Temperature must be between -50°C and 60°C.")
        except ValueError:
            print("⚠️ Invalid input. Please enter a numeric value.")

# Calculations
average_temp = sum(temperatures) / len(temperatures)
highest_temp = max(temperatures)
lowest_temp = min(temperatures)
heat_alert_days = sum(1 for t in temperatures if t > 35)

# Display results
print("\n🌡️ Weekly Temperature Summary:")
print(f"Average Temperature: {average_temp:.2f}°C")
print(f"Highest Temperature: {highest_temp}°C")
print(f"Lowest Temperature: {lowest_temp}°C")
print(f"Days with Heat Alert (>35°C): {heat_alert_days}")
```

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
fuel_consumption = []

for day in days:
    while True:
        try:
            liters = float(input(f"Enter liters of fuel consumed on {day}: "))
            if 0 < liters <= 100:
                fuel_consumption.append(liters)
                break
            else:
                print("⚠️ Please enter a positive number not exceeding 100.")
        except ValueError:
            print("⚠️ Invalid input. Please enter a numeric value.")

# Aggregations
total_fuel = sum(fuel_consumption)
average_fuel = total_fuel / len(fuel_consumption)
max_fuel = max(fuel_consumption)
max_day = days[fuel_consumption.index(max_fuel)]

# Output
print("\n⛽ Fuel Consumption Summary:")
print(f"Total fuel consumption: {total_fuel:.2f} liters")
print(f"Daily average consumption: {average_fuel:.2f} liters")
print(f"Day with highest consumption: {max_day} ({max_fuel:.2f} liters)")
```
