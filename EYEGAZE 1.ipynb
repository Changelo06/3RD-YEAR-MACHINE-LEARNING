```python
import cv2
import mediapipe as mp
import numpy as np

# Initialize MediaPipe Face Mesh
mp_face_mesh = mp.solutions.face_mesh
face_mesh = mp_face_mesh.FaceMesh(refine_landmarks=True)

# Define landmarks for the left eye and iris (MediaPipe reference)
LEFT_EYE = [362, 385, 387, 263]     # Outer eye contour
LEFT_IRIS = [474, 475, 476, 477]    # Iris landmarks

# Function to estimate gaze direction based on iris position
def estimate_gaze_direction(iris_points, eye_points):
    iris_center = np.mean(iris_points, axis=0)
    eye_center = np.mean(eye_points, axis=0)
    dx, dy = iris_center - eye_center

    threshold = 2  # sensitivity threshold

    if abs(dx) < threshold and abs(dy) < threshold:
        return "Center"
    elif dy < -threshold:
        if dx < -threshold:
            return "Northwest"
        elif dx > threshold:
            return "Northeast"
        else:
            return "North"
    elif dy > threshold:
        if dx < -threshold:
            return "Southwest"
        elif dx > threshold:
            return "Southeast"
        else:
            return "South"
    else:
        if dx < -threshold:
            return "West"
        elif dx > threshold:
            return "East"

    return "Unknown"

# Start webcam feed
cap = cv2.VideoCapture(0
                      
                      
                          )

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Flip and convert to RGB
    frame = cv2.flip(frame, 1)
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = face_mesh.process(rgb)

    if results.multi_face_landmarks:
        for face_landmarks in results.multi_face_landmarks:
            h, w, _ = frame.shape
            mesh_points = np.array([
                [int(p.x * w), int(p.y * h)]
                for p in face_landmarks.landmark
            ])

            left_eye = mesh_points[LEFT_EYE]
            left_iris = mesh_points[LEFT_IRIS]

            direction = estimate_gaze_direction(left_iris, left_eye)

            # Draw iris and direction text
            for p in left_iris:
                cv2.circle(frame, tuple(p), 2, (0, 255, 0), -1)

            cv2.putText(frame, f'Looking: {direction}', (30, 30),
                        cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2)

    cv2.imshow("Iris Gaze Tracker", frame)
    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
