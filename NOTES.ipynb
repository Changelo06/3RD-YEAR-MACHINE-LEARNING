```python
import numpy as np
import pandas as pd

def generate_battery_life_dataset(n_samples=100, random_seed=42):
    np.random.seed(random_seed)
    
    # Feature generation
    charge_cycles = np.random.randint(50, 1200, size=n_samples)
    avg_temp = np.random.uniform(20, 45, size=n_samples)
    daily_usage = np.random.uniform(1, 10, size=n_samples)
    battery_capacity = np.random.choice([3000, 4000, 4500, 5000], size=n_samples)
    charging_speed = np.random.choice([0, 1], size=n_samples)

    # Target calculation (Battery Health %)
    # Simple logic: more cycles, higher temp, and fast charging reduce battery health
    health = (
        100
        - (charge_cycles * 0.03)
        - (avg_temp - 25) * 0.7
        - (daily_usage * 0.9)
        - (charging_speed * 5)
        + (battery_capacity - 3000) * 0.002
    )

    # Add some noise
    noise = np.random.normal(0, 2, size=n_samples)
    battery_health = np.clip(health + noise, 20, 100)  # Clamp between 20% and 100%

    # Combine into DataFrame
    df = pd.DataFrame({
        'charge_cycles': charge_cycles,
        'avg_temp': np.round(avg_temp, 2),
        'daily_usage': np.round(daily_usage, 2),
        'battery_capacity': battery_capacity,
        'charging_speed': charging_speed,
        'battery_health': np.round(battery_health, 2)
    })

    return df

# Generate and preview
df_battery = generate_battery_life_dataset()
print(df_battery.head(10))
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

X = df_battery[['charge_cycles', 'avg_temp', 'daily_usage', 'battery_capacity', 'charging_speed']].to_numpy()
y = df_battery[['battery_health']].to_numpy()

# Now X and y are arrays
print("X shape:", X.shape)
print("y shape:", y.shape)
print("Sample X row:", X[0])
print("Sample y value:", y[0])
```

```python
# Determine split index
split_index = int(0.8 * len(X_scaled))  # 80% for training

# Proper train/test split without overlap
X_train = X_scaled[:split_index]
X_test = X_scaled[split_index:]
y_train = y_scaled[:split_index]
y_test = y_scaled[split_index:]
# Display number of training and test samples
print("Units for training: ", len(X_train))
print("Units for testing: ", len(X_test))
```

```python
tf.random.set_seed(42)

model = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1)
])

model.compile(
    loss=tf.keras.losses.mae,
    optimizer = tf.keras.optimizers.Adam(),
    metrics=['mae']
)

callback = tf.keras.callbacks.EarlyStopping(monitor='loss', patience=20)
model.fit(X_train, y_train, epochs=300, verbose=2, callbacks=[callback])
```

```python
loss, mae = model.evaluate(X_test, y_test)
print(f"Test MAE: {mae}")
# Evaluate the model on the test set
model.evaluate(X_test, y_test)
```
