```python
import numpy as np
import pandas as pd

# Set a random seed for reproducibility
np.random.seed(42)

# Number of samples per class
n_samples = 50

# Rose: medium petal length and width
rose_petal_length = np.random.normal(loc=5.0, scale=0.5, size=n_samples)
rose_petal_width = np.random.normal(loc=2.0, scale=0.3, size=n_samples)
rose_labels = ['Rose'] * n_samples  # String labels

# Sunflower: long petals, wide petals
sunflower_petal_length = np.random.normal(loc=7.0, scale=0.5, size=n_samples)
sunflower_petal_width = np.random.normal(loc=3.5, scale=0.4, size=n_samples)
sunflower_labels = ['Sunflower'] * n_samples

# Tulip: short petals, narrow petals
tulip_petal_length = np.random.normal(loc=4.0, scale=0.4, size=n_samples)
tulip_petal_width = np.random.normal(loc=1.5, scale=0.2, size=n_samples)
tulip_labels = ['Tulip'] * n_samples

# Combine all the data
X = np.vstack((
    np.column_stack((rose_petal_length, rose_petal_width)),
    np.column_stack((sunflower_petal_length, sunflower_petal_width)),
    np.column_stack((tulip_petal_length, tulip_petal_width))
))

y = np.array(rose_labels + sunflower_labels + tulip_labels)

# Create a DataFrame
df = pd.DataFrame(X, columns=['Petal_Length', 'Petal_Width'])
df['Label'] = y

print(df.head(0))
```

```python
import tensorflow as tf

y = pd.get_dummies(y)
y.head(1)
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
# Build our train and test sets (use random state to ensure same split as before)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
tf.random.set_seed(42) 

model_1 = tf.keras.Sequential([
    tf.keras.layers.Dense(32, activation = "relu"),
    tf.keras.layers.Dense(16, activation = "relu"),
    tf.keras.layers.Dense(3, activation = "softmax")
])

model_1.compile(
    loss="categorical_crossentropy",  # not sparse anymore
    optimizer="adam",
    metrics=["accuracy"]
)
model_1.fit(X_train, y_train, epochs = 50, validation_data = (X_test, y_test), verbose = 1)
```

```python
import numpy as np

# Example new sample (same format: petal_length, petal_width)
new_sample = np.array([[5.2, 2.1]])  # 2D array, even for a single sample

# Predict
prediction = model_1.predict(new_sample)

print("Prediction probabilities:", prediction)
```
